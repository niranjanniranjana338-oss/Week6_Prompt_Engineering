# Week 6: Prompt Engineering

This notebook shows reusable prompt styles and an optional Hugging Face generation step.

```python
import csv
from pathlib import Path

DATA = Path('../dataset')
prompt_rows = [
    ('Zero-shot', 'Explain photosynthesis in simple words for a school student.'),
    ('One-shot', 'Example: Product: Pen. Description: A smooth pen for everyday writing. Now write one for a notebook.'),
    ('Few-shot', 'Text: I love this. Sentiment: Positive\nText: This is awful. Sentiment: Negative\nText: It is okay. Sentiment:'),
    ('Role prompting', 'Act as a friendly study coach. Create a realistic one-week study plan for a college student.'),
    ('Structured reasoning', 'Help debug this Python error. Explain the key checks briefly, then provide corrected code: TypeError when adding a string and an integer.')
]
for kind, prompt in prompt_rows:
    print(f'[{kind}] {prompt}\n')
```

```python
# Optional: run a pretrained Hugging Face model. First install transformers and torch.
# The first run requires internet access to download distilgpt2.
try:
    from transformers import pipeline
    generator = pipeline('text-generation', model='distilgpt2')
    prompt = prompt_rows[0][1]
    result = generator(prompt, max_new_tokens=60, do_sample=True, temperature=0.7,
                       num_return_sequences=1, pad_token_id=generator.tokenizer.eos_token_id)
    print(result[0]['generated_text'])
except ImportError:
    print('Install dependencies with: python -m pip install -r ../requirements.txt')
```
